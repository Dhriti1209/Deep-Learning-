```python
import numpy as np
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)
```

```python
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([
    [0],
    [0],
    [0],
    [1]
])

print("Input:")
print(X)

print("\nOutput:")
print(y)
```

```python
np.random.seed(42)

weights = np.random.rand(2, 1)
bias = np.random.rand(1)

print("Initial Weights:")
print(weights)

print("\nInitial Bias:")
print(bias)
```

```python
learning_rate = 0.1
epochs = 10000
```

```python
for epoch in range(epochs):

    # Forward Propagation
    input_layer = X
    weighted_sum = np.dot(input_layer, weights) + bias
    output = sigmoid(weighted_sum)

    # Error Calculation
    error = y - output

    # Backpropagation
    d_output = error * sigmoid_derivative(output)

    # Update Weights and Bias
    weights += learning_rate * np.dot(input_layer.T, d_output)
    bias += learning_rate * np.sum(d_output)
```

```python
print("Trained Weights:")
print(weights)

print("\nTrained Bias:")
print(bias)
```

```python
print("Predicted Output after Training:")
print(output)
```
